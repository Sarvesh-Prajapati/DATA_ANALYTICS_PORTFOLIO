### The **`OS`** **Module**


Core functions (`os.name`, `os.getcwd()`, `os.chdir()`, `os.listdir()`, `os.mkdir()`, `os.makedirs()`, `os.rmdir()`, `os.remove()`, `os.rename()`), `os.path` functions (`os.path.join()`, `os.path.exists()`, `os.path.isfile()`, `os.path.isdir()`, `os.path.split()`), advanced functions (`os.environ`, `os.system()`, `os.walk()`), and its practical application in data engineering.

1.  The `os` module in Python provides a way of using operating system dependent functionality to interact with the underlying OS, abstracting away the differences between various OS like Windows, macOS, and Linux.

2.  It enables Python to perform a variety of OS tasks like file system manipulation (e.g., creating, deleting, renaming files and directories), accessing environment variables, managing processes, and executing shell commands directly from Python.

3.  In scripting, automation, and data engineering, tasks like managing files, navigating directories, and controlling system processes are crucial operations. `os` module helps accomplish them.

The `os.path` is a sub-module of the `os` module that provides common pathname manipulation functionalities. It's crucial for writing platform-independent code, as it handles the differences in path conventions between operating systems (e.g., `/` on Unix-like systems vs. `\` on Windows).

1. **`os.path.exists(path)`**: Returns `True/False` if a path (file or directory) exists/doesn't exist. Often used before performing file operations (e.g., reading, writing, deleting) to prevent `FileNotFoundError` or `DirectoryNotFoundError`.

2.  **`os.path.isfile(path)`**: Returns `True/False` if a file exists/doesn't exist. Differentiates between files and directories, useful when you specifically need to operate on files.

3.  **`os.path.isdir(path)`**: Returns `True/False` if a path points to/doesn't point to an existing directory. Similar to `isfile()`, it helps in verifying if a path points to a directory, which is important before attempting operations like listing directory contents.

4.  **`os.path.split(path)`**: Splits the pathname into a tuple, `(head, tail)`, where `tail` is the last component of the path and `head` is everything leading up to that. Useful to extract file name from a full path or getting the parent directory name.

#### **Advanced OS Module Functions**:
`os.environ` (for environment variables), `os.system()` (for executing shell commands), and `os.walk()` (for traversing directory trees).


1.  **`os.environ`**: is a dictionary-like object that represents the user's environment variables. Environment variables are dynamic named values that can affect the way running processes behave on a computer. They can store information like the search path for executables (`PATH`), the current user (`USER` or `USERNAME`), or system settings. You can access environment variables like dictionary keys (e.g., `os.environ['PATH']`). You can also add or modify environment variables, though these changes typically only last for the duration of the current Python process. Allows Python programs to retrieve system-specific configurations or user preferences without hardcoding them, making scripts more flexible and portable.

2.  **`os.system(command)`**: executes a shell command (a string) in a subshell. It's a way to run external programs or shell scripts directly from your Python code. You pass a string representing the command you want to execute (e.g., `os.system('ls -l')` on Unix-like systems, or `os.system('dir')` on Windows). The return value is the exit status of the command. Useful for tasks that are better handled by existing shell commands or external utilities (e.g., system administration, running legacy scripts, interacting with system tools).

3.  **`os.walk(top, topdown=True, onerror=None, followlinks=False)`**: `os.walk()` generates the file names in a directory tree by walking the tree either top-down or bottom-up. For each directory in the tree rooted at `top` (including `top` itself), it yields a 3-tuple: `(dirpath, dirnames, filenames)`.

        *   `dirpath`: A string, the path to the directory.
        *   `dirnames`: A list of the names of the subdirectories in `dirpath` (not including `.` or `..`).
        *   `filenames`: A list of the names of the non-directory files in `dirpath`.

    **Usage**: It's typically used in a `for` loop to iterate through the directory tree. `topdown=True` (default) means it visits the directory itself before its subdirectories. `topdown=False` processes subdirectories before the parent directory.
    
    **Why it's essential**: Indispensable for tasks that require processing all files or directories within a given root, such as indexing files, searching for specific file types, or backing up entire directory structures.

```python
import os

print("--- Demonstrating os.environ ---")

# Access and print common environment variables
# Note: Environment variable names are case-sensitive on some systems (e.g., Linux/macOS) and case-insensitive on others (e.g., Windows)
# Using .get() is safer as it won't raise an error if the variable doesn't exist

# Example 1: PATH environment variable
path_var = os.environ.get('PATH')
if path_var: print(f"PATH environment variable: {path_var[:70]}...") # Print only first 70 chars
else: print("PATH environment variable not found.")

# Example 2: HOME or HOMEPATH (for user's home directory)
home_var = os.environ.get('HOME') or os.environ.get('HOMEPATH')
if home_var: print(f"HOME/HOMEPATH environment variable: {home_var}")
else: print("HOME/HOMEPATH environment variable not found.")

# Example 3: USER (for the current user)
user_var = os.environ.get('USER') or os.environ.get('USERNAME')
if user_var: print(f"USER/USERNAME environment variable: {user_var}")
else: print("USER/USERNAME environment variable not found.")
```

```python
import os
import shutil # For rmtree to remove non-empty directories

print("\n--- Demonstrating os.walk() ---")

# Define the base directory for the demo
walk_demo_base_dir = "walk_demo_root"

# --- 4a. Create a temporary directory structure ---

print(f"Creating temporary directory structure under '{walk_demo_base_dir}'...")

# Create root directory
if not os.path.exists(walk_demo_base_dir):
    os.mkdir(walk_demo_base_dir)

# Create nested directories and files
# walk_demo_root/
#   dir_a/
#     file_a1.txt
#     dir_a1/
#       file_a11.txt
#   dir_b/
#     file_b1.csv
#   file_root.log

dirs_to_create = [
    os.path.join(walk_demo_base_dir, "dir_a"),
    os.path.join(walk_demo_base_dir, "dir_a", "dir_a1"),
    os.path.join(walk_demo_base_dir, "dir_b"),
]

files_to_create = {
    os.path.join(walk_demo_base_dir, "dir_a", "file_a1.txt"): "Content for file A1.",
    os.path.join(walk_demo_base_dir, "dir_a", "dir_a1", "file_a11.txt"): "Content for file A11.",
    os.path.join(walk_demo_base_dir, "dir_b", "file_b1.csv"): "header1,header2\nval1,val2",
    os.path.join(walk_demo_base_dir, "file_root.log"): "Log entry 1\nLog entry 2",
}

for d in dirs_to_create:
    os.makedirs(d, exist_ok=True)

for f_path, f_content in files_to_create.items():
    with open(f_path, "w") as f:
        f.write(f_content)

print("Temporary directory structure and files created.")

# --- 4b. Use os.walk() to traverse the directory structure ---

print(f"\nTraversing '{walk_demo_base_dir}' with os.walk():")
for root, dirs, files in os.walk(walk_demo_base_dir):
    print(f"\nRoot: {root}")
    print(f"  Directories: {dirs}")
    print(f"  Files: {files}")

# --- 4c. Clean up the temporary directory structure ---

print(f"\nCleaning up temporary directory '{walk_demo_base_dir}'...")
if os.path.exists(walk_demo_base_dir):
    shutil.rmtree(walk_demo_base_dir)
    print(f"Directory '{walk_demo_base_dir}' and all its contents removed.")
else:
    print(f"Directory '{walk_demo_base_dir}' does not exist.")
```

### **The `shutil`** (shell utility) **Module**


#### `os` vs `shutil`


It provides a higher-level interface for file and directory operations compared to the `os` module. While the `os` module deals with low-level, atomic interactions with the operating system (like creating a single directory or removing a single file), `shutil` builds upon these functionalities to offer more convenient and powerful operations like:

*   **Copying files and directories:** `shutil.copy()`, `shutil.copy2()`, `shutil.copyfile()`, `shutil.copytree()`
*   **Moving/renaming files and directories:** `shutil.move()`
*   **Deleting directory trees:** `shutil.rmtree()`
*   **Archiving and unarchiving:** `shutil.make_archive()`, `shutil.unpack_archive()`

`shutil` is essential for more advanced file and directory management because it simplifies complex tasks that would be cumbersome or error-prone to implement manually using only the `os` module.

For example, `os.rmdir()` can only delete empty directories, requiring a manual recursive deletion of contents before removal. In contrast, `shutil.rmtree()` can delete an entire directory tree, including all its contents, with a single function call. Similarly, `shutil.copytree()` handles the recursive copying of an entire directory structure, including permissions and metadata, which would be a multi-step process with `os` functions.

#### `shutil` Copy Functions

1.  **`shutil.copyfile(src, dst, *, follow_symlinks=True)`**:
    *   **Purpose**: Copies the contents of the file `src` to the file `dst`. The destination must be a filename, not a directory. If `dst` already exists, it will be overwritten. The file's permissions, modification times, or other metadata are *not* copied.
    *   **Use Case**: Ideal for simple content copying where only the data itself matters, and metadata preservation is not a concern (e.g., creating a backup of a file's content).

2.  **`shutil.copy(src, dst, *, follow_symlinks=True)`**:
    *   **Purpose**: Copies both the file's contents and its permission bits (mode) from `src` to `dst`. This function is similar to `copyfile()` but also attempts to preserve the file's permission mode. If `dst` specifies a directory, the file will be copied into that directory with the original filename.
    *   **Use Case**: When you need to copy a file and ensure its execute permissions (or read/write permissions) are maintained in the new location.

3.  **`shutil.copymode(src, dst)`**:
    *   **Purpose**: Copies the permission bits (mode) from `src` to `dst`. The file contents, owner, and groups are unaffected. This is useful when you want to apply the exact permissions of one file to another existing file.
    *   **Use Case**: Synchronizing permissions between files, or applying a standard set of permissions to a newly created file based on an existing template.

4.  **`shutil.copystat(src, dst)`**:
    *   **Purpose**: Copies the permission bits, last access time, last modification time, and flags from `src` to `dst`. This function does *not* copy file contents. It essentially copies all the "stat" information (metadata) from one file to another, except for owner and group.
    *   **Use Case**: Preserving timestamps for archiving, version control, or when maintaining file history is important, while not needing to copy the file's actual data.

#### `shutil.copytree` & `shutil.move` functions


Continuing with the `shutil` module, following functions provide powerful utilities for managing entire directory structures and moving files/directories.

1.  **`shutil.copytree(src, dst, symlinks=False, ignore=None, copy_function=copy2, ignore_dangling_symlinks=False, dirs_exist_ok=False)`**:
    *   **Purpose**: Recursively copies an entire directory tree from `src` to `dst`. This means it copies `src` itself and all its contents (subdirectories, files, etc.) to the `dst` location.
    *   **Parameters**:
        *   `src`: The source directory path.
        *   `dst`: The destination directory path. It must not already exist, or `dirs_exist_ok` must be `True`.
        *   `symlinks`: If `True`, symbolic links in `src` are copied as symbolic links in `dst`. If `False` (default), the contents of symbolic links are copied.
        *   `ignore`: A callable that, given a directory name and a list of its contents, returns a list of names to be ignored.
        *   `dirs_exist_ok`: If `False` (default), an error is raised if `dst` already exists. If `True`, the copy operation will continue into an existing destination directory, overwriting files where applicable.
    *   **Use Case**: Essential for creating backups of entire project structures, deploying applications, or duplicating a complex dataset directory for experimental purposes.

2.  **`shutil.move(src, dst, copy_function=copy2)`**:
    *   **Purpose**: Recursively moves a file or directory (`src`) to another location (`dst`). This is a higher-level function that handles various scenarios:
        *   If `dst` is an existing directory, `src` is moved inside `dst`.
        *   If `dst` does not exist but its parent directory does, `src` is renamed to `dst`.
        *   If `src` and `dst` are on different file systems, `src` is copied to `dst` and then removed.
    *   **Parameters**:
        *   `src`: The source file or directory path.
        *   `dst`: The destination path. This can be a new name or an existing directory.
    *   **Use Case**: Renaming files or directories, reorganizing file systems, migrating data between locations, or moving processed files to an archive directory within a data pipeline.

#### More `shutil` Functions:



1.  **`shutil.rmtree(path, ignore_errors=False, onerror=None)`**:
    *   **Purpose**: Recursively deletes an entire directory tree specified by `path`. This means it removes the directory itself and all its contents (subdirectories, files, etc.).
    *   **Caution**: This function is extremely powerful and irreversible. Use it with extreme care, as it will permanently delete files and directories without sending them to the recycle bin/trash.
    *   **Parameters**:
        *   `path`: The path to the directory tree to be removed.
        *   `ignore_errors`: If `True`, errors resulting from failed removals will be ignored. If `False` (default), such errors will be handled by the `onerror` handler.
        *   `onerror`: A callable that will be called if an error occurs. It receives the function, path, and exception information.
    *   **Use Case**: Cleaning up temporary directories, removing old data archives, or resetting a project's output directory. It's much simpler than manually removing files and then empty directories using `os` functions.

2.  **`shutil.make_archive(base_name, format, root_dir=None, base_dir=None, verbose=0, dry_run=0, owner=None, group=None, logger=None)`**:
    *   **Purpose**: Creates an archive file (like a zip or tar archive) from a source directory. This is useful for bundling multiple files and directories into a single compressed file for easier storage or transfer.
    *   **Parameters**:
        *   `base_name`: The name of the archive file to create, without the format extension (e.g., 'my_data_backup').
        *   `format`: The archive format to use (e.g., 'zip', 'tar', 'gztar', 'bztar', 'xztar').
        *   `root_dir`: The directory from which to start archiving. All paths in the archive will be relative to this directory. If `None` (default), the current working directory is used.
        *   `base_dir`: The directory that will be archived. All contents of `base_dir` will be included. If `None` (default), `root_dir` is used.
    *   **Use Case**: Creating backups of data, packaging software distributions, or bundling datasets for sharing.

3.  **`shutil.unpack_archive(filename, extract_dir=None, format=None)`**:
    *   **Purpose**: Extracts an archive file to a specified directory. This is the counterpart to `make_archive()`.
    *   **Parameters**:
        *   `filename`: The path to the archive file to unpack.
        *   `extract_dir`: The directory where the archive should be extracted. If `None` (default), the current working directory is used.
        *   `format`: The format of the archive (e.g., 'zip', 'tar'). If `None` (default), `shutil` attempts to auto-detect the format.
    *   **Use Case**: Decompressing downloaded datasets, installing software from archives, or restoring backups.

4.  **`shutil.disk_usage(path)`**:
    *   **Purpose**: Reports disk usage statistics for the given path. It returns a `namedtuple` with `total`, `used`, and `free` space in bytes.
    *   **Parameters**:
        *   `path`: The path to a directory or file on the filesystem.
    *   **Returns**: A `namedtuple` with `total`, `used`, and `free` attributes, representing bytes.
    *   **Use Case**: Monitoring disk space on servers, checking if there's enough space before a large data processing job, or optimizing storage usage in a data pipeline.
